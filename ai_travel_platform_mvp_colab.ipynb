```python

import google.generativeai as genai
import requests
import json
import pandas as pd
from datetime import datetime
import gradio as gr

genai.configure(api_key=GEMINI_API_KEY)

model = genai.GenerativeModel("gemini-1.5-flash")
```


# Weather Utility


```python

def get_weather(city):
    try:
        url = f"https://wttr.in/{city}?format=j1"
        response = requests.get(url, timeout=10)
        data = response.json()

        current = data["current_condition"][0]

        return {
            "temperature": current["temp_C"],
            "condition": current["weatherDesc"][0]["value"],
            "humidity": current["humidity"]
        }

    except Exception as e:
        return {
            "temperature": "N/A",
            "condition": "Unavailable",
            "humidity": "N/A"
        }
```


# AI Trip Planner Engine


```python

def generate_trip(
    destination,
    budget,
    days,
    interests,
    travelers
):
    weather = get_weather(destination)

    prompt = f'''
    You are an expert AI Travel Planner.

    Create a detailed travel itinerary.

    Destination: {destination}
    Budget: {budget}
    Number of Days: {days}
    Interests: {interests}
    Travelers: {travelers}

    Current Weather:
    Temperature: {weather['temperature']} C
    Condition: {weather['condition']}

    Requirements:
    - Day-wise itinerary
    - Budget-conscious recommendations
    - Hotels
    - Restaurants
    - Activities
    - Local transportation
    - Indoor alternatives if weather is bad
    - Approximate costs
    - Optimized scheduling

    Format nicely in markdown.
    '''

    response = model.generate_content(prompt)

    return response.text
```


# Conversational AI Update Engine


```python

def update_trip(existing_plan, user_request):
    prompt = f'''
    You are an AI Travel Assistant.

    Existing Trip Plan:
    {existing_plan}

    User Update Request:
    {user_request}

    Update the itinerary accordingly.
    Maintain proper formatting.
    '''

    response = model.generate_content(prompt)

    return response.text
```


# Gradio Web Application


```python

def planner_ui(destination, budget, days, interests, travelers):
    return generate_trip(
        destination,
        budget,
        days,
        interests,
        travelers
    )

planner = gr.Interface(
    fn=planner_ui,
    inputs=[
        gr.Textbox(label="Destination"),
        gr.Textbox(label="Budget"),
        gr.Slider(1, 14, value=5, step=1, label="Trip Duration (Days)"),
        gr.Textbox(label="Interests"),
        gr.Textbox(label="Travelers")
    ],
    outputs=gr.Markdown(label="Generated Itinerary"),
    title="AI Travel Planning Platform",
    description="Generate intelligent travel itineraries dynamically using Gemini AI."
)

planner.launch(share=True)
```
